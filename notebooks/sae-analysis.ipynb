```python
# Import necessary modules
import sys
sys.path.insert(0, '/home/student/s/ssahu/share/aigm-classifier/scripts')

from neural_codec_confounders import MetaEnCodecWrapper, DACWrapper
import librosa
import numpy as np
```

```python
# Load test audio file
audio_path = '/home/student/s/ssahu/share/aigm-classifier/data/trainset/real/fma_real/fma_medium/000/000002.mp3'
audio, sr = librosa.load(audio_path, sr=16000)
print(f"Loaded audio: {audio.shape}, sr={sr}")
```

```python
# Test EnCodec
encodec = MetaEnCodecWrapper(sr=16000, bandwidth=1.5)
audio_encodec = encodec.process_audio(audio)
print(f"EnCodec output shape: {audio_encodec.shape}")
```

```python
# Test DAC
dac = DACWrapper(sr=16000, model_name="16khz")
audio_dac = dac.process_audio(audio)
print(f"DAC output shape: {audio_dac.shape}")
```

```python
# Compare audio outputs
print(f"Original audio - min: {audio.min():.4f}, max: {audio.max():.4f}")
print(f"EnCodec audio - min: {audio_encodec.min():.4f}, max: {audio_encodec.max():.4f}")
print(f"DAC audio - min: {audio_dac.min():.4f}, max: {audio_dac.max():.4f}")
```
